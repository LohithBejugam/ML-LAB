```python
import pandas as pd
import math

data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}

df = pd.DataFrame(data)

print(df)
```

```python
new_brightness = 45
new_saturation = 40
```

```python
distances = []

for i in range(len(df)):
    distance = math.sqrt(
        (df.loc[i, 'Brightness'] - new_brightness) ** 2 +
        (df.loc[i, 'Saturation'] - new_saturation) ** 2
    )

    distances.append(distance)

df['Distance'] = distances

print(df)
```

```python
print("Distances from the new entry:")

for i in range(len(df)):
    print(
        "Point", i + 1,
        ":", round(df.loc[i, 'Distance'], 2)
    )
```
