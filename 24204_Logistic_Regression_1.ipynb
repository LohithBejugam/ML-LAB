```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
url = "https://stats.idre.ucla.edu/stat/data/binary.csv"

data = pd.read_csv(url)

data.head()
```

```python
print(data.shape)
print(data.info())
print(data.describe())
```

```python
print(data.isnull().sum())
```

```python
print(data['admit'].value_counts())
```

```python
model = smf.logit(
    'admit ~ gre + gpa + C(rank)',
    data=data
).fit()

print(model.summary())
```

```python
odds_ratios = np.exp(model.params)

print("Odds Ratios:")
print(odds_ratios)
```

```python
student = pd.DataFrame({
    'gre': [650],
    'gpa': [3.7],
    'rank': [2]
})

probability = model.predict(student)

print("Probability of admission:", probability.iloc[0])
```

```python
if probability.iloc[0] >= 0.5:
    print("Predicted: ADMIT")
else:
    print("Predicted: DON'T ADMIT")
```
